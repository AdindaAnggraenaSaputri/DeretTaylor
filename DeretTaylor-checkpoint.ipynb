```python
import math

# Menentukan nilai x, titik pusat a, dan nilai referensi dari math.sqrt
x = 26.0
a = 25.0
h = x - a
nilai_sebenarnya = math.sqrt(x)

# Menyimpan nilai hampiran dari P0 sampai P4
# P0(x) = 5
# P1(x) = 5 + 0.1*h
# P2(x) = 5 + 0.1*h - 0.001*h**2
# Untuk P3 dan P4, kita turunkan lebih lanjut:
# f'''(x) = (3/8) * x^(-5/2) -> f'''(25) = 3 / (8 * 3125) = 3 / 25000 = 0.00012 -> koefisien P3 = 0.00012 / 6 = 0.00002
# f""''(x) = (-15/16) * x^(-7/2) -> f""''(25) = -15 / (16 * 15625) = -3 / 80000 = -0.0000375 -> koefisien P4 = -0.0000375 / 24 = -0.0000015625

koefisien = [
    5.0,              # Koefisien untuk orde 0 (P0)
    0.1,              # Koefisien untuk orde 1 (P1)
    -0.001,           # Koefisien untuk orde 2 (P2)
    0.00002,          # Koefisien untuk orde 3 (P3)
    -0.0000015625     # Koefisien untuk orde 4 (P4)
]

p_saat_ini = 0.0
target_galat = 1e-6
orde_minimal = None

print(f"Nilai referensi sqrt(26) = {nilai_sebenarnya:.6f}\n")

# Perulangan untuk menghitung polinomial Taylor dari orde 0 hingga 4
for i in range(len(koefisien)):
    if i == 0:
        p_saat_ini = koefisien[0]
    else:
        # Menambahkan suku berikutnya ke polinomial sebelumnya
        p_saat_ini += koefisien[i] * (h ** i)
    
    # Menghitung galat absolut untuk orde ke-i
    galat = abs(nilai_sebenarnya - p_saat_ini)
    print(f"Orde P{i} = {p_saat_ini:.6f} | Galat Absolut = {galat:.8f}")
    
    # Mengecek apakah galat sudah kurang dari 10^-6
    if galat < target_galat and orde_minimal is None:
        orde_minimal = i

print(f"\nOrde minimal agar galat < 10^-6 adalah: P{orde_minimal}")
```
